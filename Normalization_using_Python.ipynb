**Normalization using Python**

Using Scikit-learn library

```python
import numpy as np
v1=np.array([[2],[1]])
print(v1)
from sklearn.preprocessing import normalize
v2=v1/np.linalg.norm(v1)
print(v2)
```

**Normailization using Python with Complex Coefficient**

```python
import numpy as np
v1=np.array([[2],[1+1j]])
print(v1)
from sklearn.preprocessing import normalize
v2=v1/np.linalg.norm(v1)
print(v2)
```
